```python

#1. TO IMPLEMENT BREADTH FIRST SEARCH (BFS) USING PYTHON

from collections import deque

# Function to perform BFS
def bfs(graph, start):
    visited = set()         #to keep the track of the visited nodes
    queue = deque([start])   #initialize queue with starting node
    visited.add(start)

    print("BFS Traversal:", end=" ")

    while queue:
        vertex = queue.popleft()
        print(vertex, end=" ")

        for neighbor in graph[vertex]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)


graph = {}
n = int(input("Enter the number of vertices: "))

for i in range(n):
    vertex = input(f"Enter vertex {i+1}: ")
    neighbor = input(
        f"Enter adjacent vertices of {vertex} (space-separated): "
    ).split()

    graph[vertex] = neighbor

start = input("Enter the starting vertex: ")

print("\nGraph:", graph)
bfs(graph, start)
```
